### Config

```python
import tiktoken
from importlib.metadata import version
import tiktoken
print("tiktoken version:", version("tiktoken"))
print(tiktoken.list_encoding_names())
```

```python
DATA_PATH = "data/short_story.txt"
```

### Read text

```python
text = None
with open(DATA_PATH, 'r') as f:
    text = f.read()
```

### TikToken usage

```python
tokenizer = tiktoken.get_encoding("o200k_base")
print(tokenizer.special_tokens_set)
```

```python
tok_ids = tokenizer.encode(text, allowed_special={"<|endoftext|>"})
tok_ids[:10]
```

```python
tokenizer.decode(tok_ids)[:100]
```

### Multiple sequences

```python
sent1 = "Hello, do you like tea?"
sent2 = "In the sunlit terraces of the palace."
sents = sent1 + "<|endoftext|>"+sent2
sents
```

```python
tok_ids = tokenizer.encode(sents, allowed_special={"<|endoftext|>"})
print("TOKEN IDs: ", tok_ids)
print("\nTOKENS: ", [tokenizer.decode([tok_id]) for tok_id in tok_ids])
print("\nDECODED: ", tokenizer.decode(tok_ids))
```

```python
import torch
from util import prepare_dataset
```

### Config

```python
DATA_PATH = "data/short_story.txt"
CONTEXT_SIZE = 8
STRIDE = 4
BATCH_SIZE = 16
EMB_DIM = 32
```

```python
tokenizer, loader = prepare_dataset(
    DATA_PATH, context_size=CONTEXT_SIZE, stride=STRIDE, batch_size=BATCH_SIZE
)
```

```python
x, y = next(iter(loader))
print(
    "X=\n'",
    tokenizer.decode(x[0].numpy()),
    "'\nY=\n'",
    tokenizer.decode(y[0].numpy()),
    "'",
)
print(x.shape)
print(y.shape)
```

### Token Embeddings

```python
tokenizer.n_vocab
```

```python
emb_layer = torch.nn.Embedding(tokenizer.n_vocab, EMB_DIM)
```

```python
emb_layer.weight.shape
```

```python
emb_layer(torch.tensor([120]))
```

```python
emb_layer(torch.tensor([120])).shape
```

### Positional Embeddings

```python
pos_emb_layer = torch.nn.Embedding(tokenizer.n_vocab, EMB_DIM)
```

```python
pos_emb_input = torch.arange(CONTEXT_SIZE)
```

```python
pos_emb = pos_emb_layer(pos_emb_input)
```

```python
pos_emb.shape
```

### Compute input embeddings

```python
input_emb = emb_layer(x) + pos_emb
```
